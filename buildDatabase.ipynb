```python
from typing import Tuple
from datamodels.Tupleable import Tupleable

def generateSqlString(dbTable: str, objTuple: Tuple) -> str:

    # Creating placeholder string.
    placeholders = ", ".join(["?"] * len(objTuple))

    sql = f"INSERT INTO {dbTable} VALUES({placeholders});"

    return sql

def save(conn: sqlite3.Connection, cursor: sqlite3.Cursor, dbTable: str, objId: str, tupleable: Tupleable):

    # Generating DB-calibrated tuple.
    objTuple = (objId,) + tupleable.toTuple()

    sql = generateSqlString(dbTable, objTuple)

    # Execute and commit changes to DB.
    cursor.execute(sql, objTuple)
    conn.commit()

def saveWithParentUUID(conn: sqlite3.Connection, cursor: sqlite3.Cursor, dbTable: str, objId: str, tupleable: Tupleable, parentUUID: str):

    # NOTE: This assumes that the foreign UUID is the to be placed at index 1 (second position) of the tuple.
    # Generating DB-calibrated tuple.
    objTuple = (objId, parentUUID) + tupleable.toTuple()

    sql = generateSqlString(dbTable, objTuple)

    # Execute and commit changes to DB.
    cursor.execute(sql, objTuple)
    conn.commit()
```

```python
from datamodels.Race import Race

race = Race(2026, 1, 1)
raceUUID = str(uuid.uuid4())
save(conn, cursor, "race", raceUUID, race)

for event in race.schedule:
    eventUUID = str(uuid.uuid4())
    saveWithParentUUID(conn, cursor, "event", eventUUID, event, raceUUID)

feed = race.feed
feedUUID = str(uuid.uuid4())
saveWithParentUUID(conn, cursor, "feed", feedUUID, feed, raceUUID)

for result in feed.results:
    resultUUID = str(uuid.uuid4())
    saveWithParentUUID(conn, cursor, "result", resultUUID, result, feedUUID)

for caution in feed.cautionSegments:
    cautionUUID = str(uuid.uuid4())
    saveWithParentUUID(conn, cursor, "caution", cautionUUID, caution, feedUUID)

for leader in feed.raceLeaders:
    leaderUUID = str(uuid.uuid4())
    saveWithParentUUID(conn, cursor, "leader", leaderUUID, leader, feedUUID)

for stage in feed.stages:
    stageUUID = str(uuid.uuid4())
    saveWithParentUUID(conn, cursor, "stage", stageUUID, stage, feedUUID)

    for stageFinisher in stage.results:
        stageFinisherUUID = str(uuid.uuid4())
        saveWithParentUUID(conn, cursor, "stageFinisher", stageFinisherUUID, stageFinisher, stageUUID)

# Handling pitReports.
pitReportUUID = str(uuid.uuid4())
# Generating DB-calibrated tuple.
objTuple = (pitReportUUID, feedUUID) + (str(feed.pitReports),)

sql = generateSqlString("pitReport", objTuple)

# Execute and commit changes to DB.
cursor.execute(sql, objTuple)
conn.commit()

for pitStop in race.pitStops:
    pitStopUUID = str(uuid.uuid4())
    saveWithParentUUID(conn, cursor, "pitStop", pitStopUUID, pitStop, raceUUID)

# for lap in race.driverLaps:
#     lapUUID = str(uuid.uuid4())
#     saveWithParentUUID(conn, cursor, "lap", lapUUID, lap, raceUUID)

for session in race.sessions:
    sessionUUID = str(uuid.uuid4())
    saveWithParentUUID(conn, cursor, "session", sessionUUID, session, raceUUID)

    for sessionResult in session.results:
        sessionResultUUID = str(uuid.uuid4())
        saveWithParentUUID(conn, cursor, "sessionResult", sessionResultUUID, sessionResult, sessionUUID)
```
