```python
import sqlite3

# Initializing database and config.
conn = sqlite3.connect("2022_26_cup.db")
cursor = conn.cursor()
```

```python
# Implementing NasPy schema.
# Race schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "race(uuid TEXT PRIMARY KEY,"
               "season INTEGER,"
               "seriesID INTEGER,"
               "round INTEGER,"
               "raceID INTEGER,"
               "raceName TEXT,"
               "raceTypeID INTEGER,"
               "restrictorPlate NUMERIC,"
               "trackID INTEGER,"
               "trackName TEXT,"
               "dateScheduled NUMERIC,"
               "raceDate NUMERIC,"
               "qualifyingDate NUMERIC,"
               "tuneInDate NUMERIC,"
               "scheduledDistance REAL,"
               "actualDistance REAL,"
               "scheduledLaps INTEGER,"
               "actualLaps INTEGER,"
               "stage1Laps INTEGER,"
               "stage2Laps INTEGER,"
               "stage3Laps INTEGER,"
               "carCount INTEGER,"
               "poleWinnerDriverID INTEGER,"
               "poleWinnerSpeed REAL,"
               "numberOfLeadChanges INTEGER,"
               "numberOfLeaders INTEGER,"
               "numberOfCautions INTEGER,"
               "numberOfCautionLaps INTEGER,"
               "averageSpeed REAL,"
               "totalRaceTime NUMERIC,"
               "marginOfVictory REAL,"
               "racePurse REAL,"
               "raceComments TEXT,"
               "attendance INTEGER,"
               "infractions TEXT,"
               "radioBroadcaster TEXT,"
               "tvBroadcaster TEXT,"
               "satelliteRadioBroadcaster TEXT,"
               "masterRaceID INTEGER,"
               "inspectionComplete NUMERIC,"
               "playoffRound INTEGER,"
               "isQualifyingRace NUMERIC,"
               "qualifyingRaceNo INTEGER,"
               "qualifyingRaceID INTEGER,"
               "hasQualifying NUMERIC,"
               "winnerDriverID INTEGER,"
               "poleWinnerLapTime NUMERIC"
               ")")

# Event schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "event("
               "uuid TEXT PRIMARY KEY,"
               "raceUUID TEXT,"
               "eventName TEXT,"
               "notes TEXT,"
               "startTime NUMERIC,"
               "runType INTEGER,"
               "FOREIGN KEY(raceUUID) REFERENCES race(uuid)"
               ")")

# Feed schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "feed(uuid TEXT PRIMARY KEY,"
               "raceUUID TEXT,"
               "stage4Laps INTEGER,"
               "FOREIGN KEY(raceUUID) REFERENCES race(uuid)"
               ")")

# Result schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "result(uuid TEXT PRIMARY KEY,"
               "feedUUID TEXT,"
               "resultID INTEGER,"
               "finishingPosition INTEGER,"
               "startingPositioning INTEGER,"
               "carNumber TEXT,"
               "driverFullname TEXT,"
               "driverID INTEGER,"
               "driverHometown TEXT,"
               "hometownCity TEXT,"
               "hometownState TEXT,"
               "hometownCountry TEXT,"
               "teamID INTEGER,"
               "teamName TEXT,"
               "qualifyingOrder INTEGER,"
               "qualifyingPosition INTEGER,"
               "qualifyingSpeed REAL,"
               "lapsLed INTEGER,"
               "timesLed INTEGER,"
               "carMake TEXT,"
               "carModel TEXT,"
               "sponsor TEXT,"
               "pointsEarned INTEGER,"
               "playoffPointsEarned INTEGER,"
               "lapsCompleted INTEGER,"
               "finishingStatus TEXT,"
               "winnings REAL,"
               "seriesID INTEGER,"
               "raceSeason INTEGER,"
               "raceID INTEGER,"
               "ownerFullname TEXT,"
               "crewChiefID INTEGER,"
               "crewChiefFullname TEXT,"
               "pointsPosition INTEGER,"
               "pointsDelta INTEGER,"
               "ownerID INTEGER,"
               "officialCarNumber TEXT,"
               "disqualified NUMERIC,"
               "diffLaps INTEGER,"
               "diffTime INTEGER,"
               "pitBox INTEGER,"
               "FOREIGN KEY(feedUUID) REFERENCES feed(uuid)"
               ")")

# Caution schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "caution(uuid TEXT PRIMARY KEY,"
               "feedUUID TEXT,"
               "raceID INTEGER,"
               "startLap INTEGER,"
               "endLap INTEGER,"
               "reason TEXT,"
               "comment TEXT,"
               "beneficiary TEXT,"
               "flagState INTEGER,"
               "FOREIGN KEY(feedUUID) REFERENCES feed(uuid)"
               ")")

#Leader schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "leader(uuid TEXT PRIMARY KEY,"
               "feedUUID TEXT,"
               "startLap INTEGER,"
               "endLap INTEGER,"
               "carNumber TEXT,"
               "raceID INTEGER,"
               "FOREIGN KEY(feedUUID) REFERENCES feed(uuid)"
               ")")

# PitStop schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "pitStop(uuid TEXT PRIMARY KEY,"
               "raceUUID TEXT,"
               "vehicleNumber TEXT,"
               "driverName TEXT,"
               "vehicleManufacturer TEXT,"
               "leaderLap INTEGER,"
               "lapCount INTEGER,"
               "pitInFlagStatus INTEGER,"
               "pitOutFlagStatus INTEGER,"
               "pitInRaceTime REAL,"
               "pitOutRaceTime REAL,"
               "totalDuration REAL,"
               "boxStopRaceTime REAL,"
               "boxLeaveRaceTime REAL,"
               "pitStopDuration REAL,"
               "inTravelDuration REAL,"
               "outTravelDuration REAL,"
               "pitStopType TEXT,"
               "leftFrontTireChanged NUMERIC,"
               "leftRearTireChanged NUMERIC,"
               "rightFrontTireChanged NUMERIC,"
               "rightRearTireChanged NUMERIC,"
               "previousLapTime REAL,"
               "nextLapTime REAL,"
               "pitInRank INTEGER,"
               "pitOutRank INTEGER,"
               "positionsGainedLost INTEGER,"
               "FOREIGN KEY(raceUUID) REFERENCES race(uuid)"
               ")")

# Lap schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "lap(uuid TEXT PRIMARY KEY,"
               "raceUUID TEXT,"
               "lapNumber INTEGER,"
               "lapTime REAL,"
               "lapSpeed REAL,"
               "runningPosition INTEGER,"
               "FOREIGN KEY(raceUUID) REFERENCES race(uuid)"
               ")")

# Stage schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "stage(uuid TEXT PRIMARY KEY,"
               "feedUUID TEXT,"
               "stageNumber INTEGER,"
               "FOREIGN KEY(feedUUID) REFERENCES feed(uuid)"
               ")")

# StageFinisher schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "stageFinisher(uuid TEXT PRIMARY KEY,"
               "stageUUID TEXT,"
               "driverFullName TEXT,"
               "driverID INTEGER,"
               "carNumber TEXT,"
               "finishingPosition INTEGER,"
               "stagePoints INTEGER,"
               "FOREIGN KEY(stageUUID) REFERENCES stage(uuid)"
               ")")

# Session schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "session(uuid TEXT PRIMARY KEY,"
               "raceUUID TEXT,"
               "weekendRunID INTEGER,"
               "raceID INTEGER,"
               "timingRunID INTEGER,"
               "runType INTEGER,"
               "runName TEXT,"
               "runDate NUMERIC,"
               "runDateUTC NUMERIC,"
               "FOREIGN KEY(raceUUID) REFERENCES race(uuid)"
               ")")

# SessionResult schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "sessionResult(uuid TEXT PRIMARY KEY,"
               "sessionUUID TEXT,"
               "runID INTEGER,"
               "carNumber TEXT,"
               "vehicleNumber TEXT,"
               "manufacturer TEXT,"
               "driverID INTEGER,"
               "driverName TEXT,"
               "finishingPosition INTEGER,"
               "bestLapTime REAL,"
               "bestLapNumber INTEGER,"
               "lapsCompleted INTEGER,"
               "comment TEXT,"
               "deltaLeader REAL,"
               "disqualified NUMERIC,"
               "FOREIGN KEY (sessionUUID) REFERENCES session(uuid)"
               ")")

# PitReport schema.
cursor.execute("CREATE TABLE IF NOT EXISTS "
               "pitReport(uuid TEXT PRIMARY KEY,"
               "feedUUID TEXT,"
               "val TEXT,"
               "FOREIGN KEY(feedUUID) REFERENCES feed(uuid)"
               ")")
```
