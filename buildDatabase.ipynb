```python
import uuid
import sqlite3

def configDatabase(dbName: str) -> (sqlite3.Cursor, sqlite3.Connection):
    # Initializing database and config.
    conn = sqlite3.connect(dbName)
    dbCursor = conn.cursor()

    return dbCursor, conn
```

```python
# Implementing NasPy schema.
def initializeSchema(dbCursor: sqlite3.Cursor):
    # Race schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "race(uuid TEXT PRIMARY KEY,"
                   "season INTEGER,"
                   "seriesID INTEGER,"
                   "round INTEGER,"
                   "raceID INTEGER,"
                   "raceName TEXT,"
                   "raceTypeID INTEGER,"
                   "restrictorPlate NUMERIC,"
                   "trackID INTEGER,"
                   "trackName TEXT,"
                   "dateScheduled NUMERIC,"
                   "raceDate NUMERIC,"
                   "qualifyingDate NUMERIC,"
                   "tuneInDate NUMERIC,"
                   "scheduledDistance REAL,"
                   "actualDistance REAL,"
                   "scheduledLaps INTEGER,"
                   "actualLaps INTEGER,"
                   "stage1Laps INTEGER,"
                   "stage2Laps INTEGER,"
                   "stage3Laps INTEGER,"
                   "carCount INTEGER,"
                   "poleWinnerDriverID INTEGER,"
                   "poleWinnerSpeed REAL,"
                   "numberOfLeadChanges INTEGER,"
                   "numberOfLeaders INTEGER,"
                   "numberOfCautions INTEGER,"
                   "numberOfCautionLaps INTEGER,"
                   "averageSpeed REAL,"
                   "totalRaceTime NUMERIC,"
                   "marginOfVictory REAL,"
                   "racePurse REAL,"
                   "raceComments TEXT,"
                   "attendance INTEGER,"
                   "infractions TEXT,"
                   "radioBroadcaster TEXT,"
                   "tvBroadcaster TEXT,"
                   "satelliteRadioBroadcaster TEXT,"
                   "masterRaceID INTEGER,"
                   "inspectionComplete NUMERIC,"
                   "playoffRound INTEGER,"
                   "isQualifyingRace NUMERIC,"
                   "qualifyingRaceNo INTEGER,"
                   "qualifyingRaceID INTEGER,"
                   "hasQualifying NUMERIC,"
                   "winnerDriverID INTEGER,"
                   "poleWinnerLapTime NUMERIC"
                   ")")

    # Event schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "event("
                   "uuid TEXT PRIMARY KEY,"
                   "raceUUID TEXT,"
                   "eventName TEXT,"
                   "notes TEXT,"
                   "startTime NUMERIC,"
                   "runType INTEGER,"
                   "FOREIGN KEY(raceUUID) REFERENCES race(uuid)"
                   ")")

    # Event index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS eventIndex ON event(raceUUID)")

    # Feed schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "feed(uuid TEXT PRIMARY KEY,"
                   "raceUUID TEXT,"
                   "stage4Laps INTEGER,"
                   "FOREIGN KEY(raceUUID) REFERENCES race(uuid)"
                   ")")

    # Feed index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS feedIndex ON feed(raceUUID)")

    # Result schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "result(uuid TEXT PRIMARY KEY,"
                   "feedUUID TEXT,"
                   "resultID INTEGER,"
                   "finishingPosition INTEGER,"
                   "startingPositioning INTEGER,"
                   "carNumber TEXT,"
                   "driverFullname TEXT,"
                   "driverID INTEGER,"
                   "driverHometown TEXT,"
                   "hometownCity TEXT,"
                   "hometownState TEXT,"
                   "hometownCountry TEXT,"
                   "teamID INTEGER,"
                   "teamName TEXT,"
                   "qualifyingOrder INTEGER,"
                   "qualifyingPosition INTEGER,"
                   "qualifyingSpeed REAL,"
                   "lapsLed INTEGER,"
                   "timesLed INTEGER,"
                   "carMake TEXT,"
                   "carModel TEXT,"
                   "sponsor TEXT,"
                   "pointsEarned INTEGER,"
                   "playoffPointsEarned INTEGER,"
                   "lapsCompleted INTEGER,"
                   "finishingStatus TEXT,"
                   "winnings REAL,"
                   "seriesID INTEGER,"
                   "raceSeason INTEGER,"
                   "raceID INTEGER,"
                   "ownerFullname TEXT,"
                   "crewChiefID INTEGER,"
                   "crewChiefFullname TEXT,"
                   "pointsPosition INTEGER,"
                   "pointsDelta INTEGER,"
                   "ownerID INTEGER,"
                   "officialCarNumber TEXT,"
                   "disqualified NUMERIC,"
                   "diffLaps INTEGER,"
                   "diffTime INTEGER,"
                   "pitBox INTEGER,"
                   "FOREIGN KEY(feedUUID) REFERENCES feed(uuid)"
                   ")")

    # Result index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS resultIndex ON result(feedUUID)")

    # Caution schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "caution(uuid TEXT PRIMARY KEY,"
                   "feedUUID TEXT,"
                   "raceID INTEGER,"
                   "startLap INTEGER,"
                   "endLap INTEGER,"
                   "reason TEXT,"
                   "comment TEXT,"
                   "beneficiary TEXT,"
                   "flagState INTEGER,"
                   "FOREIGN KEY(feedUUID) REFERENCES feed(uuid)"
                   ")")

    # Caution index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS cautionIndex ON caution(feedUUID)")

    #Leader schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "leader(uuid TEXT PRIMARY KEY,"
                   "feedUUID TEXT,"
                   "startLap INTEGER,"
                   "endLap INTEGER,"
                   "carNumber TEXT,"
                   "raceID INTEGER,"
                   "FOREIGN KEY(feedUUID) REFERENCES feed(uuid)"
                   ")")

    # Leader index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS leaderIndex ON leader(feedUUID)")

    # PitStop schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "pitStop(uuid TEXT PRIMARY KEY,"
                   "raceUUID TEXT,"
                   "vehicleNumber TEXT,"
                   "driverName TEXT,"
                   "vehicleManufacturer TEXT,"
                   "leaderLap INTEGER,"
                   "lapCount INTEGER,"
                   "pitInFlagStatus INTEGER,"
                   "pitOutFlagStatus INTEGER,"
                   "pitInRaceTime REAL,"
                   "pitOutRaceTime REAL,"
                   "totalDuration REAL,"
                   "boxStopRaceTime REAL,"
                   "boxLeaveRaceTime REAL,"
                   "pitStopDuration REAL,"
                   "inTravelDuration REAL,"
                   "outTravelDuration REAL,"
                   "pitStopType TEXT,"
                   "leftFrontTireChanged NUMERIC,"
                   "leftRearTireChanged NUMERIC,"
                   "rightFrontTireChanged NUMERIC,"
                   "rightRearTireChanged NUMERIC,"
                   "previousLapTime REAL,"
                   "nextLapTime REAL,"
                   "pitInRank INTEGER,"
                   "pitOutRank INTEGER,"
                   "positionsGainedLost INTEGER,"
                   "FOREIGN KEY(raceUUID) REFERENCES race(uuid)"
                   ")")

    # PitStop index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS pitStopIndex ON pitStop(raceUUID)")

    # Stage schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "stage(uuid TEXT PRIMARY KEY,"
                   "feedUUID TEXT,"
                   "stageNumber INTEGER,"
                   "FOREIGN KEY(feedUUID) REFERENCES feed(uuid)"
                   ")")

    # Stage index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS stageIndex ON stage(feedUUID)")

    # StageFinisher schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "stageFinisher(uuid TEXT PRIMARY KEY,"
                   "stageUUID TEXT,"
                   "driverFullName TEXT,"
                   "driverID INTEGER,"
                   "carNumber TEXT,"
                   "finishingPosition INTEGER,"
                   "stagePoints INTEGER,"
                   "FOREIGN KEY(stageUUID) REFERENCES stage(uuid)"
                   ")")

    # StageFinisher index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS stageFinisherIndex ON stageFinisher(stageUUID)")

    # DriverLapsBatch schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "driverLapsBatch(uuid TEXT PRIMARY KEY,"
                   "raceUUID TEXT,"
                   "carNumber TEXT,"
                   "fullName TEXT,"
                   "manufacturer TEXT,"
                   "runningPosition INTEGER,"
                   "driverID INTEGER,"
                   "FOREIGN KEY(raceUUID) REFERENCES race(uuid)"
                   ")")

    # DriverLapsBatch index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS driverLapsBatchIndex ON driverLapsBatch(raceUUID)")

    # Lap schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "lap(uuid TEXT PRIMARY KEY,"
                   "batchUUID TEXT,"
                   "lapNumber INTEGER,"
                   "lapTime REAL,"
                   "lapSpeed REAL,"
                   "runningPosition INTEGER,"
                   "FOREIGN KEY(batchUUID) REFERENCES driverLapsBatch(uuid)"
                   ")")

    # Lap index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS lapIndex ON lap(batchUUID)")

    # Session schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "session(uuid TEXT PRIMARY KEY,"
                   "raceUUID TEXT,"
                   "weekendRunID INTEGER,"
                   "raceID INTEGER,"
                   "timingRunID INTEGER,"
                   "runType INTEGER,"
                   "runName TEXT,"
                   "runDate NUMERIC,"
                   "runDateUTC NUMERIC,"
                   "FOREIGN KEY(raceUUID) REFERENCES race(uuid)"
                   ")")

    # Session index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS sessionIndex ON session(raceUUID)")

    # SessionResult schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "sessionResult(uuid TEXT PRIMARY KEY,"
                   "sessionUUID TEXT,"
                   "runID INTEGER,"
                   "carNumber TEXT,"
                   "vehicleNumber TEXT,"
                   "manufacturer TEXT,"
                   "driverID INTEGER,"
                   "driverName TEXT,"
                   "finishingPosition INTEGER,"
                   "bestLapTime REAL,"
                   "bestLapSpeed REAL,"
                   "bestLapNumber INTEGER,"
                   "lapsCompleted INTEGER,"
                   "comment TEXT,"
                   "deltaLeader REAL,"
                   "disqualified NUMERIC,"
                   "FOREIGN KEY (sessionUUID) REFERENCES session(uuid)"
                   ")")

    # SessionResult index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS sessionResultIndex ON sessionResult(sessionUUID)")

    # PitReport schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "pitReport(uuid TEXT PRIMARY KEY,"
                   "feedUUID TEXT,"
                   "val TEXT,"
                   "FOREIGN KEY(feedUUID) REFERENCES feed(uuid)"
                   ")")

    # PitReport index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS pitReportIndex ON pitReport(feedUUID)")

    # LapNote schema.
    dbCursor.execute("CREATE TABLE IF NOT EXISTS "
                   "lapNote(uuid TEXT PRIMARY KEY,"
                   "raceUUID TEXT,"
                   "raceSeason INTEGER,"
                   "seriesID INTEGER,"
                   "raceID INTEGER,"
                   "lapNumber INTEGER,"
                   "flagState INTEGER,"
                   "note TEXT,"
                   "noteID INTEGER,"
                   "driverIDs TEXT,"
                   "FOREIGN KEY (raceUUID) REFERENCES race(uuid)"
                   ")")

    # LapNote index.
    dbCursor.execute("CREATE INDEX IF NOT EXISTS lapNoteIndex ON lapNote(raceUUID)")
```

```python
from datamodels.Feed import Feed
from typing import Tuple
from datamodels.Tupleable import Tupleable

def generateSqlString(dbTable: str, objTuple: Tuple) -> str:

    # Creating placeholder string.
    placeholders = ", ".join(["?"] * len(objTuple))

    sql = f"INSERT INTO {dbTable} VALUES({placeholders});"

    return sql

def save(conn: sqlite3.Connection, dbCursor: sqlite3.Cursor, dbTable: str, objId: str, tupleable: Tupleable):

    # Generating DB-calibrated tuple.
    objTuple = (objId,) + tupleable.toTuple()

    sql = generateSqlString(dbTable, objTuple)

    # Execute and commit changes to DB.
    dbCursor.execute(sql, objTuple)
    conn.commit()

def saveWithParentUUID(conn: sqlite3.Connection, dbCursor: sqlite3.Cursor, dbTable: str, objId: str, tupleable: Tupleable, parentUUID: str):

    # NOTE: This assumes that the foreign UUID is the to be placed at index 1 (second position) of the tuple.
    # Generating DB-calibrated tuple.
    objTuple = (objId, parentUUID) + tupleable.toTuple()

    sql = generateSqlString(dbTable, objTuple)

    # Execute and commit changes to DB.
    dbCursor.execute(sql, objTuple)
    conn.commit()

def savePitReports(conn: sqlite3.Connection, dbCursor: sqlite3.Cursor, feedUUID: str, feed: Feed):
    # Handling pitReports.
    pitReportUUID = str(uuid.uuid4())
    # Generating DB-calibrated tuple.
    objTuple = (pitReportUUID, feedUUID) + (str(feed.pitReports),)

    sql = generateSqlString("pitReport", objTuple)

    # Execute and commit changes to DB.
    dbCursor.execute(sql, objTuple)
    conn.commit()
```

```python
from helpers import Parsers
from datamodels.Race import Race

def buildDatabase(conn: sqlite3.Connection, dbCursor: sqlite3.Cursor, seasons: list, seriesIDs: list):

    for season in seasons:
        for seriesID in seriesIDs:
            for race in range(36):

                print(f"""\n\n========== ROUND {race} ==========\n\n""")

                race = Race(season, seriesID, race, includeExhibitions=False)
                raceUUID = str(uuid.uuid4())
                save(conn, dbCursor, "race", raceUUID, race)
                print(f"[RACE] Successfully persisted {race.season} {race.raceName} - {race.raceID}.")

                for event in race.schedule:
                    eventUUID = str(uuid.uuid4())
                    saveWithParentUUID(conn, dbCursor, "event", eventUUID, event, raceUUID)
                print(f"[EVENT] Successfully persisted events for {race.season} {race.raceName} - {race.raceID}.")

                feed = race.feed
                feedUUID = str(uuid.uuid4())
                saveWithParentUUID(conn, dbCursor, "feed", feedUUID, feed, raceUUID)
                print(f"[FEED] Successfully persisted feed for {race.season} {race.raceName} - {race.raceID}.")

                for result in feed.results:
                    resultUUID = str(uuid.uuid4())
                    saveWithParentUUID(conn, dbCursor, "result", resultUUID, result, feedUUID)
                print(f"[RESULT] Successfully persisted race results for {race.season} {race.raceName} - {race.raceID}.")

                for caution in feed.cautionSegments:
                    cautionUUID = str(uuid.uuid4())
                    saveWithParentUUID(conn, dbCursor, "caution", cautionUUID, caution, feedUUID)
                print(f"[CAUTION] Successfully persisted cautions for {race.season} {race.raceName} - {race.raceID}.")

                for leader in feed.raceLeaders:
                    leaderUUID = str(uuid.uuid4())
                    saveWithParentUUID(conn, dbCursor, "leader", leaderUUID, leader, feedUUID)
                print(f"[LEADER] Successfully persisted leaders for {race.season} {race.raceName} - {race.raceID}.")

                for stage in feed.stages:
                    stageUUID = str(uuid.uuid4())
                    saveWithParentUUID(conn, dbCursor, "stage", stageUUID, stage, feedUUID)

                    for stageFinisher in stage.results:
                        stageFinisherUUID = str(uuid.uuid4())
                        saveWithParentUUID(conn, dbCursor, "stageFinisher", stageFinisherUUID, stageFinisher, stageUUID)
                    print(f"[STAGE FINISHER] Successfully persisted stage finishers for {race.season} {race.raceName} Stage {stage.stageNumber} - {race.raceID}.")
                print(f"[STAGE] Successfully persisted stages for {race.season} {race.raceName} - {race.raceID}.")

                savePitReports(conn, dbCursor, feedUUID, feed)
                print(f"[PIT REPORT] Successfully persisted pit reports for {race.season} {race.raceName} - {race.raceID}.")

                for pitStop in race.pitStops:
                    pitStopUUID = str(uuid.uuid4())
                    saveWithParentUUID(conn, dbCursor, "pitStop", pitStopUUID, pitStop, raceUUID)
                print(f"[PIT STOP] Successfully persisted pit stops for {race.season} {race.raceName} - {race.raceID}.")

                for batch in race.driverLapsBatch:
                    batchUUID = str(uuid.uuid4())
                    saveWithParentUUID(conn, dbCursor, "driverLapsBatch", batchUUID, batch, raceUUID)

                    for lap in batch.laps:
                        lapUUID = str(uuid.uuid4())
                        saveWithParentUUID(conn, dbCursor, "lap", lapUUID, lap, batchUUID)
                    print(f"[LAP] Successfully persisted driver laps for {race.season} {race.raceName} - {race.raceID}.")
                print(f"[BATCH] Successfully persisted driver lap batches for {race.season} {race.raceName} - {race.raceID}.")

                for session in race.sessions:
                    sessionUUID = str(uuid.uuid4())
                    saveWithParentUUID(conn, dbCursor, "session", sessionUUID, session, raceUUID)

                    for sessionResult in session.results:
                        sessionResultUUID = str(uuid.uuid4())
                        saveWithParentUUID(conn, dbCursor, "sessionResult", sessionResultUUID, sessionResult, sessionUUID)
                    print(f"[SESSION RESULT] Successfully persisted session results for {race.season} {race.raceName} - {race.raceID}.")
                print(f"[SESSION] Successfully persisted sessions for {race.season} {race.raceName} - {race.raceID}.")
                lapNoteList = Parsers.buildLapNoteList(2026, 1, race.raceID)

                for lapNote in lapNoteList:
                    lapNoteUUID = str(uuid.uuid4())
                    saveWithParentUUID(conn, dbCursor, "lapNote", lapNoteUUID, lapNote, raceUUID)
                print(f"[LAP NOTE] Successfully persisted lap notes for {race.season} {race.raceName} - {race.raceID}.")

        print(f"\n[{season}] Completed season data persistence.\n")
```

```python
cursor, conn = configDatabase("22-26Cup.db")
initializeSchema(cursor)
buildDatabase(conn, cursor, [2022, 2023, 2024, 2025, 2026], [1])
```
